```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Embedding, Flatten, Concatenate, GlobalAveragePooling1D, Dropout, BatchNormalization
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam
```

```python
# Prepare input features and labels
X_words = combined_df['word'].values
X_labels = combined_df['label'].values
y = combined_df['genre'].values

# Step 2: Encode labels and target variable
# Encode 'label' and 'genre' as integers
label_encoder = LabelEncoder()
X_labels_encoded = label_encoder.fit_transform(X_labels)

genre_encoder = LabelEncoder()
y_encoded = genre_encoder.fit_transform(y)
y_categorical = to_categorical(y_encoded)

# Tokenize and pad 'word' input
word_tokenizer = Tokenizer()
word_tokenizer.fit_on_texts(X_words)
X_words_seq = word_tokenizer.texts_to_sequences(X_words)
X_words_padded = pad_sequences(X_words_seq, maxlen=10)

# Step 3: Neural Network Architecture
# Define input layers
word_input = Input(shape=(10,))
label_input = Input(shape=(1,))

# Embedding for word input
word_embedding = Embedding(input_dim=len(word_tokenizer.word_index) + 1, output_dim=128)(word_input)
word_embedding = Flatten()(word_embedding)

# Embedding for label input
label_embedding = Embedding(input_dim=len(label_encoder.classes_), output_dim=10)(label_input)
label_embedding = Flatten()(label_embedding)

# Combine the embeddings
combined = Concatenate()([word_embedding, label_embedding])

# # Neural network layers
dense_layer = Dense(128, activation='relu')(combined)
dense_layer = Dropout(0.5)(dense_layer)
dense_layer = Dropout(0.5)(dense_layer)
dense_layer = Dropout(0.5)(dense_layer)
dense_layer = Dropout(0.5)(dense_layer)
dense_layer = Dropout(0.5)(dense_layer)
dense_layer = Dropout(0.5)(dense_layer)

dense_layer = Dense(128, activation='relu')(combined)


output_layer = Dense(len(genre_encoder.classes_), activation='softmax')(dense_layer)

# Build and compile model
optimizer = Adam(learning_rate=0.01)
model = Model(inputs=[word_input, label_input], outputs=output_layer)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Step 4: Prepare training and testing data
X_train_words, X_test_words, X_train_labels, X_test_labels, y_train, y_test = train_test_split(
    X_words_padded, X_labels_encoded, y_categorical, test_size=0.25, random_state=42
)

# Train the model
model.fit([X_train_words, X_train_labels], y_train, epochs=35, batch_size=40, validation_split=0.2)

# Evaluate the model
accuracy = model.evaluate([X_test_words, X_test_labels], y_test)[1]
print(f"Model Accuracy: {accuracy:.2f}")
```

```python
def predict_genre(word, label):
    # Preprocess the input
    word_seq = word_tokenizer.texts_to_sequences([word])
    word_padded = pad_sequences(word_seq, maxlen=10)

    # Encode the label
    label_encoded = label_encoder.transform([label])  # Use transform instead of fit_transform

    # Make the prediction
    prediction = model.predict([word_padded, np.array([label_encoded])])

    # Get the predicted genre index
    predicted_index = np.argmax(prediction)

    # Decode the predicted genre back to its original label
    predicted_genre = genre_encoder.inverse_transform([predicted_index])[0]

    return predicted_genre
```
